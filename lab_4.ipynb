## Implement Linear Search and Time Measurement

### Subtask:
Implement the Linear Search algorithm as described in the notebook. Write code to measure the time taken to search for an element for different input sizes (n) and store these times.


**Reasoning**:
The first step is to import the necessary libraries, define the `linear_search` function, and set up the experimental framework to measure its performance for different input sizes. This code block will implement the linear search logic, generate test data, and collect timing measurements as per the instructions.



```python
import time
import random
import matplotlib.pyplot as plt

def linear_search(arr, target):
    """
    Implements the linear search algorithm.
    Returns the index of the target if found, otherwise -1.
    """
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1

# Generate a range of input sizes n
n_values = [n * 1000 for n in range(1, 11)]  # From 1000 to 10000

times = []
num_runs = 5 # Number of runs for averaging time

print("Measuring performance for Linear Search...")
for n in n_values:
    total_time_for_n = 0
    for _ in range(num_runs):
        # Create a list of n random integers
        # Using a fixed range for values to avoid potential issues with very large numbers
        random_list = [random.randint(0, n * 2) for _ in range(n)]

        # Choose an element to search for
        # 50% chance to search for an existing element, 50% chance for a non-existing one
        if random.random() < 0.5:
            target_element = random.choice(random_list) # Element known to be in the list
        else:
            target_element = random.randint(n * 2 + 1, n * 3) # Element likely not in the list

        start_time = time.time()
        linear_search(random_list, target_element)
        end_time = time.time()
        total_time_for_n += (end_time - start_time)

    average_time = total_time_for_n / num_runs
    times.append(average_time)
    print(f"  n = {n}: Average time = {average_time:.6f} seconds")

# Store results for later plotting (optional, but good practice)
# For now, we will use the generated lists directly for plotting in the next step.
print("Performance measurement completed.")

# Plotting will be done in the next step to keep cell size manageable.
```

**Reasoning**:
The previous step successfully measured the time taken for linear search across different input sizes. The next logical step, as per the problem description, is to visualize this relationship by plotting a graph of time taken versus the number of elements (n).



```python
plt.figure(figsize=(10, 6))
plt.plot(n_values, times, marker='o', linestyle='-', color='b')
plt.title('Linear Search Time Complexity')
plt.xlabel('Number of Elements (n)')
plt.ylabel('Average Time (seconds)')
plt.grid(True)
plt.xticks(n_values)
plt.tight_layout()
plt.show()
```

## Implement Recursive Binary Search and Time Measurement

### Subtask:
Implement the recursive Binary Search algorithm as described in the notebook. Write code to measure the time taken to search for an element for different input sizes (n) and store these times. Ensure the list is sorted before applying Binary Search.


**Reasoning**:
First, I need to define the recursive binary search function and set up the experimental framework to measure its performance for different input sizes. This includes generating sorted test data and collecting timing measurements, following the instructions provided.



```python
def recursive_binary_search(arr, target, low, high):
    """
    Implements the recursive binary search algorithm.
    Returns the index of the target if found, otherwise -1.
    """
    if high >= low:
        mid = low + (high - low) // 2

        # If element is present at the middle itself
        if arr[mid] == target:
            return mid

        # If element is smaller than mid, then it can only be present in left subarray
        elif arr[mid] > target:
            return recursive_binary_search(arr, target, low, mid - 1)

        # Else the element can only be present in right subarray
        else:
            return recursive_binary_search(arr, target, mid + 1, high)

    else:
        # Element is not present in the array
        return -1

binary_search_times = []

print("\nMeasuring performance for Recursive Binary Search...")
for n in n_values:
    total_time_for_n = 0
    for _ in range(num_runs):
        # Create a list of n random integers and sort it for binary search
        random_list = sorted([random.randint(0, n * 2) for _ in range(n)])

        # Choose an element to search for
        # 50% chance to search for an existing element, 50% chance for a non-existing one
        if random.random() < 0.5:
            target_element = random.choice(random_list) # Element known to be in the list
        else:
            target_element = random.randint(n * 2 + 1, n * 3) # Element likely not in the list

        start_time = time.time()
        recursive_binary_search(random_list, target_element, 0, len(random_list) - 1)
        end_time = time.time()
        total_time_for_n += (end_time - start_time)

    average_time = total_time_for_n / num_runs
    binary_search_times.append(average_time)
    print(f"  n = {n}: Average time = {average_time:.9f} seconds")

print("Recursive Binary Search performance measurement completed.")
```

**Reasoning**:
The previous step successfully measured the time taken for recursive binary search across different input sizes. The next logical step, as per the problem description, is to visualize this relationship by plotting a graph of time taken versus the number of elements (n).



```python
plt.figure(figsize=(10, 6))
plt.plot(n_values, binary_search_times, marker='o', linestyle='-', color='r')
plt.title('Recursive Binary Search Time Complexity')
plt.xlabel('Number of Elements (n)')
plt.ylabel('Average Time (seconds)')
plt.grid(True)
plt.xticks(n_values)
plt.tight_layout()
plt.show()
```

## Implement Insertion Sort and Time Measurement

### Subtask:
Implement the Insertion Sort algorithm as described in the notebook. Write code to measure the time taken to sort elements for different input sizes (n) and store these times.

#### Instructions
1. Define a Python function named `insertion_sort` that takes a list as input and sorts it in place using the insertion sort algorithm.
2. Initialize an empty list, for example `insertion_sort_times`, to store the average time taken for each input size.
3. Iterate through the `n_values` (input sizes) that were used for previous experiments.
4. Inside the loop, for each `n`:
    a. Generate a list of `n` random integers. It is important that the list is unsorted initially.
    b. Measure the execution time of the `insertion_sort` function for `num_runs` times, calculating the average time for the current `n`.
    c. Append this average time to the `insertion_sort_times` list.
    d. Print the `n` value and its corresponding average time.

**Reasoning**:
Following the instructions in the markdown block, I will now implement the `insertion_sort` function and set up the experimental framework to measure its performance for different input sizes. This includes generating unsorted test data and collecting timing measurements, as specified in the subtask.



```python
def insertion_sort(arr):
    """
    Implements the insertion sort algorithm.
    Sorts the array in place.
    """
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        # Move elements of arr[0..i-1], that are greater than key,
        # to one position ahead of their current position
        while j >= 0 and key < arr[j]:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key

insertion_sort_times = []

print("\nMeasuring performance for Insertion Sort...")
for n in n_values:
    total_time_for_n = 0
    for _ in range(num_runs):
        # Create an unsorted list of n random integers
        random_list = [random.randint(0, n * 2) for _ in range(n)]

        start_time = time.time()
        insertion_sort(random_list)
        end_time = time.time()
        total_time_for_n += (end_time - start_time)

    average_time = total_time_for_n / num_runs
    insertion_sort_times.append(average_time)
    print(f"  n = {n}: Average time = {average_time:.6f} seconds")

print("Insertion Sort performance measurement completed.")
```

**Reasoning**:
The previous step successfully measured the time taken for Insertion Sort across different input sizes. The next logical step, as per the problem description and the pattern established for the other algorithms, is to visualize this relationship by plotting a graph of time taken versus the number of elements (n).



```python
plt.figure(figsize=(10, 6))
plt.plot(n_values, insertion_sort_times, marker='o', linestyle='-', color='g')
plt.title('Insertion Sort Time Complexity')
plt.xlabel('Number of Elements (n)')
plt.ylabel('Average Time (seconds)')
plt.grid(True)
plt.xticks(n_values)
plt.tight_layout()
plt.show()
```

## Implement Pattern Search (String Matching)

### Subtask:
Implement a function `search(pat, txt)` that prints all occurrences of a given pattern `pat` in a text `txt`. Assume `n > m`, where `n` is the length of `txt` and `m` is the length of `pat`.

#### Instructions
1. Define a Python function named `search` that takes two string arguments: `pat` (the pattern) and `txt` (the text).
2. Inside the function, implement the algorithm described in the notebook's Prob-3 section:
    a. Initialize an empty list (e.g., `occurrences`) to store the starting indices where the pattern is found.
    b. Loop through the `txt` string from the first character up to the point where the remaining text length is at least `m` (length of `pat`).
    c. For each position `i`, extract a substring of `txt` of length `m` starting at `i`.
    d. Compare this substring with `pat`.
    e. If they match, append the current index `i` to the `occurrences` list.
    f. After the loop, print a message indicating whether the pattern was found and, if so, print the list of `occurrences` indices.
3. Test the `search` function with a few example `pat` and `txt` pairs to demonstrate its correctness.

**Reasoning**:
Following the instructions in the markdown block, I will now implement the `search` function for pattern matching. This includes defining the function, iterating through the text, comparing substrings, and reporting the occurrences as specified in the subtask.



```python
def search(pat, txt):
    """
    Searches for all occurrences of a pattern in a text.
    Prints the indices where the pattern is found.
    Assumes len(txt) > len(pat).
    """
    n = len(txt)
    m = len(pat)
    occurrences = []

    # Loop through the text up to the point where a full pattern can still fit
    for i in range(n - m + 1):
        # Extract a substring of length m starting at i
        substring = txt[i : i + m]

        # Compare the substring with the pattern
        if substring == pat:
            occurrences.append(i)

    if occurrences:
        print(f"Pattern '{pat}' found at indices: {occurrences}")
    else:
        print(f"Pattern '{pat}' not found in the text.")

# Test cases
print("\n--- Testing Pattern Search ---")

# Test case 1: Pattern found multiple times
text1 = "ABABDABACDABABCABAB"
pattern1 = "ABABCABAB"
print(f"Text: '{text1}'")
print(f"Pattern: '{pattern1}'")
search(pattern1, text1)

# Test case 2: Pattern found once
text2 = "GEEKSFORGEEKS"
pattern2 = "GEEKS"
print(f"\nText: '{text2}'")
print(f"Pattern: '{pattern2}'")
search(pattern2, text2)

# Test case 3: Pattern not found
text3 = "AAAAA"
pattern3 = "BBA"
print(f"\nText: '{text3}'")
print(f"Pattern: '{pattern3}'")
search(pattern3, text3)

# Test case 4: Pattern at the beginning and end
text4 = "ABCDEFABC"
pattern4 = "ABC"
print(f"\nText: '{text4}'")
print(f"Pattern: '{pattern4}'")
search(pattern4, text4)
```
