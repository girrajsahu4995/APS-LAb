# 1. Growth Of Function
Time Complexity/Order of Growth defines the amount of time taken by any program with respect to the size of the input. Time Complexity specifies how the program would behave as the order of size of input is increased. So, Time Complexity is just a function of size of its input. We will go through some of basic and most common time complexities such as:

- *Constant Time Complexity O(1) : constant running time*
- *Linear Time Complexity O(n) : linear running time*
- *Logarithmic Time Complexity O(log n) : logarithmic running time*
- *Log-Linear Time Complexity O(n log n) : log-­linear running time*
- *Polynomial Time Complexity O(n^c) : polynomial running time (c is a constant)*
- *Exponential Time Complexity O(c^n) : exponential running time (c is a constant being raised to a power based on size of input)*

```python
import matplotlib.pyplot as plt

# Prob_2: Take another example and write the code that runs in a fixed amount of time or has a fixed number of steps of execution no matter what the size of input has constant time complexity.
# similar to example-1, also plot the graph (eg. even_odd)

def check_even_odd(n):
    # This function performs a single modulo operation and comparison,
    # which takes constant time regardless of the input 'n'.
    return "Even" if n % 2 == 0 else "Odd"

n_values = list(range(1, 101))
time_values = []

for n in n_values:
    check_even_odd(n)
    time_values.append(1) # Constant time operation, so time is always 1 unit

plt.figure()
plt.plot(n_values, time_values)
plt.xlabel("Input size n")
plt.ylabel("Time")
plt.title("O(1) Constant Time Complexity for Even/Odd Check")
plt.show()
```

#Linear Time Complexity O(n) : linear running time
- *An algorithm is said to have a linear time complexity when the running time increases linearly with the length of the input. When the function involves checking all the values in input data, with this order O(n).*

```python
def linear_time_sum(numbers):
    total = 0
    for value in numbers:
        total += value
    return total

n_values = list(range(1, 101))
time_values = []

for n in n_values:
    data = list(range(n))
    linear_time_sum(data)
    time_values.append(n)

plt.figure()
plt.plot(n_values, time_values)
plt.xlabel("Input size n")
plt.ylabel("Time")
plt.title("O(n) Linear Time Complexity")
plt.show()
```

#Logarithmic Time Complexity O(log n) : logarithmic running time
- *An algorithm is said to have a logarithmic time complexity when it reduces the size of the input data in each step. This indicates that the number of operations is not the same as the input size. The number of operations gets reduced as the input size increases. Algorithms are found in binary tree or binary search functions. This involves the search of a given value in an array by splitting the array into two and starting searching in one split. This ensures the operation is not done on every element of the data.*

```python
import matplotlib.pyplot as plt
import math

# Prob_4 Take Suitable example and write function that runs in logarithic time. plot graph as well similar to above example.

def binary_search_steps(arr, target):
    low = 0
    high = len(arr) - 1
    steps = 0
    while low <= high:
        steps += 1
        mid = (low + high) // 2
        if arr[mid] == target:
            return steps
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return steps # Target not found, return total steps taken

# For plotting, we need an array to search in and a target.
# The number of steps for binary search is approximately log2(n).
# We'll simulate input sizes and record the approximate log(n) steps.
n_values = []
time_values = []

# Let's consider array sizes that are powers of 2 for clearer logarithmic behavior
for i in range(1, 15):
    n = 2**i # Input size n
    arr = list(range(n))

    # We need a target value to search for. For consistency, let's always search for the last element.
    # The actual search time will vary slightly based on the target position,
    # but the order of growth remains O(log n).
    target = n - 1
    steps = binary_search_steps(arr, target)

    n_values.append(n)
    time_values.append(steps)

plt.figure(figsize=(10, 6))
plt.plot(n_values, time_values, label='Binary Search Steps (Actual)')

# Plot theoretical O(log n) for comparison
theoretical_log_n = [math.log2(n) for n in n_values]
plt.plot(n_values, theoretical_log_n, linestyle='--', color='red', label='Theoretical O(log n)')

plt.xlabel("Input size n")
plt.ylabel("Number of Steps")
plt.title("O(log n) Logarithmic Time Complexity (Binary Search)")
plt.legend()
plt.grid(True)
plt.show()
```

#*Log-Linear Time Complexity O(n log n) : log-­linear running time*


* When we call a Logarithmic Time Algorithm inside a loop, it would
result into a Log-Linear Time Complexity program.




```python
import matplotlib.pyplot as plt
import math

# Prob_5 Take Suitable example and write function that runs in log-linear time. plot the graph as well, similar to above example.

def merge_sort_operations(n):
    # Simulate operations for Merge Sort
    # Merge Sort has a time complexity of O(n log n).
    # For simplicity in plotting, we will use n * log2(n) as the number of operations.
    # Handle n=0 or n=1 case where log is undefined or 0
    if n <= 1:
        return n
    return n * math.log2(n)

n_values = list(range(1, 201, 5)) # Input sizes from 1 to 200, step of 5
time_values = []

for n in n_values:
    time_values.append(merge_sort_operations(n))

plt.figure(figsize=(10, 6))
plt.plot(n_values, time_values, label='Simulated Merge Sort Operations (O(n log n))')

plt.xlabel("Input size n")
plt.ylabel("Number of Operations")
plt.title("O(n log n) Log-Linear Time Complexity (Merge Sort)")
plt.legend()
plt.grid(True)
plt.show()
```

### Polynomial Time Complexity O(n^c) : polynomial running time (c is a constant)

* When the computation time increases as function of N raised to some power, N being the size of input. Such a code has Polynomial Time complexity.




```python
import matplotlib.pyplot as plt

# Prob_6 Take Suitable example and write function that runs in polynomial time. plot the graph as well, similar to above example

def polynomial_time_example(n):
    count = 0
    for i in range(n):
        for j in range(n):
            count += 1
    return count

n_values = list(range(1, 101, 2)) # Input sizes from 1 to 100
time_values = []

for n in n_values:
    time_values.append(polynomial_time_example(n)) # The number of operations is n^2

plt.figure(figsize=(10, 6))
plt.plot(n_values, time_values, label='Simulated O(n^2) Operations')

plt.xlabel("Input size n")
plt.ylabel("Number of Operations")
plt.title("O(n^2) Polynomial Time Complexity")
plt.legend()
plt.grid(True)
plt.show()
```

#Exponential Time Complexity O(c^n) : exponential running time (c is a constant being raised to a power based on size of input)

* When the computation time of a code increases as function of X^N, N being the size of input. Such a code has Polynomial Time Complexity.




```python
import matplotlib.pyplot as plt

# Prob_7 Take Suitable example and write function that runs in exponential time. plot the graph as well, similar to above example

# This dictionary will store the number of operations for each call to fib_exp
# to simulate and plot the growth. In a real scenario, this would be actual time.
operation_counts = {}

def fib_exp(n):
    if n in operation_counts:
        operation_counts[n] += 1
    else:
        operation_counts[n] = 1

    if n <= 1:
        return n
    else:
        return fib_exp(n-1) + fib_exp(n-2)

def get_total_operations(n):
    global operation_counts
    operation_counts = {} # Reset for each n
    fib_exp(n)
    return sum(operation_counts.values())

n_values = list(range(1, 16)) # Exponential growth becomes very large quickly
time_values = []

for n in n_values:
    time_values.append(get_total_operations(n))

plt.figure(figsize=(10, 6))
plt.plot(n_values, time_values, label='Simulated O(2^n) Operations (Naive Fibonacci)')

plt.xlabel("Input size n")
plt.ylabel("Number of Operations")
plt.title("O(2^n) Exponential Time Complexity (Naive Fibonacci)")
plt.legend()
plt.grid(True)
plt.show()
```
